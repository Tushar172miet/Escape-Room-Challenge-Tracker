Q2 – Multiple Session Management

```python
# Q2 - Multiple Session Management

room_limits = {
    "Lost Laboratory": 60,
    "Ancient Vault": 50,
    "Space Station": 45
}

sessions = []
unique_rooms = set()

fast_count = 0
escaped_count = 0
exceeded_count = 0

while True:
    team = input("\nEnter team name (DONE to stop): ").strip().title()

    if team == "Done":
        break

    room = input("Enter room theme: ").strip().title()

    if room not in room_limits:
        print("Invalid room theme!")
        continue

    try:
        players = int(input("Enter number of players: "))
        raw_time = float(input("Enter raw completion time: "))
        hints = int(input("Enter hints used: "))
    except ValueError:
        print("Please enter valid numbers!")
        continue

    if players < 2 or players > 6:
        print("Players must be between 2 and 6.")
        continue

    if raw_time <= 0 or hints < 0:
        print("Invalid time or hints!")
        continue

    penalty = hints * 3
    adjusted_time = raw_time + penalty
    limit = room_limits[room]

    if adjusted_time <= 0.70 * limit:
        outcome = "FAST ESCAPE"
        fast_count += 1
    elif adjusted_time <= limit:
        outcome = "ESCAPED"
        escaped_count += 1
    else:
        outcome = "TIME EXCEEDED"
        exceeded_count += 1

    record = {
        "team": team,
        "room": room,
        "players": players,
        "adjusted_time": adjusted_time,
        "outcome": outcome
    }

    sessions.append(record)
    unique_rooms.add(room)

    print("\nOutcome:", outcome)
    print("Adjusted Time:", adjusted_time)

print("\n===== FINAL SESSION REPORT =====")
print("Total Sessions:", len(sessions))
print("Fast Escapes:", fast_count)
print("Escaped:", escaped_count)
print("Time Exceeded:", exceeded_count)
print("Unique Rooms:", unique_rooms)

if sessions:
    best = min(sessions, key=lambda x: x["adjusted_time"])
    print("Best Team:", best["team"])
    print("Best Time:", best["adjusted_time"])
```

Object-Oriented Programming


```python
# Q3 - Object-Oriented Programming

class EscapeSession:

    venue_name = "Mystery Escape Arena"

    def __init__(self, session_id, team_name, room_theme,
                 players, raw_time, hints_used, room_limit):

        self.session_id = session_id
        self.team_name = team_name
        self.room_theme = room_theme
        self.players = players
        self.raw_time = raw_time
        self.hints_used = hints_used
        self.room_limit = room_limit

    def adjusted_time(self):
        return self.raw_time + (self.hints_used * 3)

    def outcome(self):
        time = self.adjusted_time()

        if time <= 0.70 * self.room_limit:
            return "FAST ESCAPE"
        elif time <= self.room_limit:
            return "ESCAPED"
        else:
            return "TIME EXCEEDED"

    def display_summary(self):
        print("\nSession ID:", self.session_id)
        print("Team:", self.team_name)
        print("Room:", self.room_theme)
        print("Players:", self.players)
        print("Adjusted Time:", self.adjusted_time())
        print("Outcome:", self.outcome())
        print("Venue:", self.venue_name)

    def __str__(self):
        return (self.team_name + " - " +
                self.room_theme + " - " + self.outcome())

session1 = EscapeSession(
    101, "Code Breakers", "Lost Laboratory",
    4, 35, 2, 60
)

session2 = EscapeSession(
    102, "Puzzle Masters", "Ancient Vault",
    5, 43, 1, 50
)


session1.display_summary()
session2.display_summary()

print("\nString Representation:")
print(session1)
print(session2)
```

Functions and Modules

```python
%%writefile escape_utils.py

def adjusted_time(raw_time, hints_used, penalty_per_hint=3):
    return raw_time + (hints_used * penalty_per_hint)


def classify_outcome(time, room_limit):
    if time <= 0.70 * room_limit:
        return "FAST ESCAPE"
    elif time <= room_limit:
        return "ESCAPED"
    else:
        return "TIME EXCEEDED"


def session_summary(team, room, time, outcome):
    return f"Team: {team} | Room: {room} | Time: {time} | Outcome: {outcome}"



def summarize_sessions(sessions):
    counts = {
        "FAST ESCAPE": 0,
        "ESCAPED": 0,
        "TIME EXCEEDED": 0
    }

    for session in sessions:
        counts[session["outcome"]] += 1

    best_team = min(sessions, key=lambda x: x["adjusted_time"])["team"] if sessions else "None"

    return counts, best_team

def total_hints(*hint_counts):
    return sum(hint_counts)
```

```python
# Q4 - Functions and Modules

from escape_utils import (
    adjusted_time,
    classify_outcome,
    session_summary,
    summarize_sessions,
    total_hints
)

import math

sessions = [
    {
        "team": "Code Breakers",
        "room": "Lost Laboratory",
        "adjusted_time": adjusted_time(35, 2),
        "outcome": classify_outcome(adjusted_time(35, 2), 60)
    },
    {
        "team": "Puzzle Masters",
        "room": "Ancient Vault",
        "adjusted_time": adjusted_time(43, 1),
        "outcome": classify_outcome(adjusted_time(43, 1), 50)
    }
]

for session in sessions:
    print(session_summary(
        session["team"],
        session["room"],
        session["adjusted_time"],
        session["outcome"]
    ))

    slots = math.ceil(session["adjusted_time"] / 15)
    print("Billing Slots:", slots)

counts, best_team = summarize_sessions(sessions)

print("\nOutcome Counts:", counts)
print("Best Team:", best_team)

print("Total Hints:", total_hints(2, 1, 3, 2))
```

File Handling and Exception Handling

```python
%%writefile -a escape_utils.py

# Q5 - Save session records to a text file
def save_session(session):
    record = (
        f"{session['id']}|{session['team']}|{session['room']}|"
        f"{session['players']}|{session['raw_time']}|{session['hints']}|"
        f"{session['adjusted_time']}|{session['outcome']}\n"
    )

    with open("escape_sessions.txt", "a") as file:
        file.write(record)


# Read saved session records
def read_sessions():
    try:
        with open("escape_sessions.txt", "r") as file:
            print("\n===== SAVED SESSION RECORDS =====")

            for line in file:
                fields = line.strip().split("|")

                if len(fields) != 8:
                    print("Error: Malformed record detected.")
                    continue

                try:
                    int(fields[0])
                    int(fields[3])
                    float(fields[4])
                    int(fields[5])
                    float(fields[6])
                except ValueError:
                    print("Error: Invalid data in record.")
                    continue

                print(
                    "ID:", fields[0],
                    "| Team:", fields[1],
                    "| Room:", fields[2],
                    "| Adjusted Time:", fields[6],
                    "| Outcome:", fields[7]
                )

    except FileNotFoundError:
        print("Error: Session file not found.")
```
